#NLP phase-1:Text Processing
#text->Cleaning->Tokenisation->stop words removal->streaming/Lemmetization->Vectorization

```python
text="Iam learning python programming and Python is fast!!..."
print("Original text:",text)
```

```python
#step-1(Convert to lowercase)
text=text.lower()
#Python !=python
print("Lowercase:", text)
```

```python
#step-2 Remove punctuation -!@#$&*_;'
import re
text=re.sub(r'[^\w\s]','',text)
print("Without punctuation: ",text)
```

```python
#Tokenisation(Split sentence into words)
tokens=text.split()
print("Tokens",tokens)
```

```python
#Step-4 :Remove stopwords-is, am, are, the, a,an
stopwords=["a","am","is","i","the","and"]
filtered_tokens=[]
for word in tokens:
  if word not in stopwords:
    filtered_tokens.append(word)
print("After removing stopwords: ",filtered_tokens)
```

```python
def stem(word):
  if word.endswith("ing"):
    return word[:-3]
  return word
stemmed_words=[]
for word in filtered_tokens:
  stemmed_words.append(stem(word))
print("After stemming: ",stemmed_words)
```

NLP Phase-2: Word to vector

```python
#NLP Phase-2: Word to vector
#why vectors?
#ML models only understand numbers
#Bag of Words(BoW)
```

```python
#Will convert text->count
from sklearn.feature_extraction.text import CountVectorizer
```

```python
sentences=["iam learning python","we are happy today","iam sad today"]
```

```python
vectorizer=CountVectorizer()
X=vectorizer.fit_transform(sentences)
print("Vocabulary: ",vectorizer.get_feature_names_out())
#learns vocabulary
#convert sentences to numbers
```

```python
print ("Vectors")
print(X.toarray())
#each row is a sentence
#each column is a word
#value is a count
```

```python
#tokenization
from nltk.tokenize import word_tokenize
#stopwordslist
from nltk.corpus import stopwords
#stemming
from nltk.stem import PorterStemmer
#lemmatization
from nltk.stem import WordNetLemmatizer
```

```python
raw_text="Iam learning Python programming, Python is gratest language..!"
#lower case
raw_text=raw_text.lower()
print("After lowercaqse: ",raw_text)
```

```python
#step-2 tokenisation
print(raw_text.split())
```

```python
tokens=word_tokenize(raw_text)
print("Tokens: ",tokens)
```

```python
#step-3 remove stopwords
english_stopwords=stopwords.words("english")
```

```python
filtered_tokens=[]
for word in tokens:
  if word not in filtered_tokens:
    filtered_tokens.append(word)
print("filtered_tokens: ",filtered_tokens)
```

```python
#step-4 remove punctuations
import string
punctuations=string.punctuation
print("Punctuations available: ",punctuations)
```

```python
clean_tokens=[word for word in filtered_tokens if word not in punctuations]
print("After removing tokens: ",clean_tokens)
```

```python
stem=PorterStemmer()
stem.stem("bought")
```

```python
wnet=WordNetLemmatizer()
wnet.lemmatize("went","v")
```

```python
lemmatized_words=[]
for word in clean_tokens:
  lemmatized_words.append(wnet.lemmatize(word,"v"))
print("After lemmatizaton: ",lemmatized_words)
```

```python
final_tokens=[]
for word in lemmatized_words:
  if word.isalpha():
    final_tokens.append(word)
print("Final Tokens: ",final_tokens)
```

```python
cleaned_text=" ".join(final_tokens)
print("Cleand text: ",cleaned_text)
```

```python
sentences = [
    "i am learning python and python",
    "we are happy today",
    "i am sad today"
]
```

```python
def preprocess_text(documents):
    stop_words = set(stopwords.words('english'))
    lemmatizer = WordNetLemmatizer()

    cleaned_docs = []

    for sentence in documents:
        words = sentence.lower().split()

        filtered_words = []
        for word in words:
            if word not in stop_words:
                lemma = lemmatizer.lemmatize(word)
                filtered_words.append(lemma)

        cleaned_docs.append(" ".join(filtered_words))

    return cleaned_docs
```

```python
cleaned_sentences=preprocess_text(sentences)
print(cleaned_sentences)
```
